```python
df.groupby("Category")["Sales"].sum()
```

```python
df.groupby("Category")["Sales"].sum().plot(kind="bar")

plt.title("Sales by Category")
plt.xlabel("Category")
plt.ylabel("Total Sales")

plt.show()
```

```python
df.groupby("Region")["Sales"].sum().plot(kind="bar")

plt.title("Sales by Region")
plt.xlabel("Region")
plt.ylabel("Total Sales")

plt.show()
```

```python
df.groupby("Segment")["Sales"].sum().plot(kind="bar")

plt.title("Sales by Customer Segment")
plt.xlabel("Segment")
plt.ylabel("Total Sales")
plt.show()
```

```python
top_products = df.groupby("Product Name")["Sales"].sum().sort_values(ascending=False).head(10)

top_products.plot(kind="barh")

plt.title("Top 10 Products by Sales")
plt.xlabel("Total Sales")
plt.ylabel("Product Name")
plt.show()
```

```python
df["Order Date"] = pd.to_datetime(df["Order Date"], dayfirst=True)
```

```python
df["Month"] = df["Order Date"].dt.month
```

```python
df.groupby("Month")["Sales"].sum().plot(kind="line", marker="o")

plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Total Sales")
plt.show()
```

### Insight: Monthly Sales Trend

Sales fluctuated throughout the year, with some months generating significantly higher revenue than others.

This trend may indicate seasonal buying behavior, promotional periods, or increased customer demand during certain months.
